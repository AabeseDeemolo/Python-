#### String Slincing[satrt,end,step]

```python
Example_string = "I Done My First Python Task"
print(type(Example_string))

print("length : ",len(Example_string))
print("1 :",Example_string[0:])     # starts from 0th position
print("2 :",Example_string[0:27])    # starts from 0th position ends with 27th position
print("3 :",Example_string[0:27:2])  # step include 
print("4 :",Example_string[0:7])
print("5 :",Example_string[16:27])
print("6 :",Example_string[16:27:3])

print("\n   Reverse slincing")          #---------   Nagative Indexing Allowed   -------------
print("7 :",Example_string[:-4])     # start from 0 ends with backward -4
print("8 :",Example_string[-12:-4])     # starting index and ending index nagative (backward)
print("9 :",Example_string[-12:-4:2])     # nagative indexes with step

print("\n   Reverse a string use slicing")
print("10 :",Example_string[::-1])     # reverse the string using slice
print("11 :",Example_string[::-2])      # reverse the string step by 

```

#### 23 . startswith()    - > Return True if S starts with the specified prefix, False otherwise. With optional start, test S beginning at that position. With optional end, stop comparing S at that position. prefix can also be a tuple of strings to try.
