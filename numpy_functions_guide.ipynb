# NumPy Functions Guide (Beginner Friendly)
NumPy is a Python library for working with numbers and arrays.
Each section has a short explanation and an example you can run.

**Install command (run once in terminal):** `pip install numpy`
**In Jupyter/Colab:** `!pip install numpy`

```python
import numpy as np
print(np.__version__)
```

## 1. Creating Arrays

```python
a = np.array([1, 2, 3, 4])            # from a list
b = np.array([[1, 2, 3], [4, 5, 6]])  # 2D array
print(a)
print(b)
```

```python
print(np.zeros((2, 3)))       # all zeros
print(np.ones((2, 3)))        # all ones
print(np.full((2, 2), 7))     # all same value
print(np.eye(3))              # identity matrix
print(np.empty((2, 2)))       # empty (random leftover values)
```

```python
print(np.arange(0, 10, 2))      # start, stop, step -> [0 2 4 6 8]
print(np.linspace(0, 1, 5))     # 5 equal points between 0 and 1
print(np.logspace(0, 2, 3))     # 10^0, 10^1, 10^2
print(np.zeros_like(b))         # zeros with same shape as b
print(np.ones_like(b))
print(np.full_like(b, 9))
```

## 2. Array Information

```python
x = np.array([[1, 2, 3], [4, 5, 6]])
print(x.shape)     # (rows, columns)
print(x.ndim)      # number of dimensions
print(x.size)      # total items
print(x.dtype)     # data type
print(x.itemsize)  # bytes per item
print(len(x))      # length of first axis
```

```python
y = x.astype(float)   # change data type
print(y.dtype)
print(np.shape(x), np.size(x), np.ndim(x))
```

## 3. Reshaping

```python
r = np.arange(12)
print(r.reshape(3, 4))        # 3 rows, 4 columns
print(r.reshape(2, -1))       # -1 means "work it out for me"
print(x.flatten())            # make 1D (copy)
print(x.ravel())              # make 1D (view)
print(x.T)                    # transpose
print(np.transpose(x))
print(np.expand_dims(a, axis=0).shape)   # add a dimension
print(np.squeeze(np.zeros((1, 3, 1))).shape)  # remove size-1 dimensions
print(np.swapaxes(x, 0, 1))
```

## 4. Indexing and Slicing

```python
m = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(m[0, 1])      # row 0, column 1 -> 20
print(m[1])         # whole row 1
print(m[:, 2])      # whole column 2
print(m[0:2, 1:])   # sub-array
print(m[-1])        # last row
print(m[::2])       # every 2nd row
```

```python
print(m[[0, 2]])            # fancy indexing: rows 0 and 2
print(m[m > 40])            # boolean indexing
print(np.take(m, [0, 4]))   # take by flat index
```

## 5. Math Operations

```python
p = np.array([1, 2, 3])
q = np.array([4, 5, 6])
print(p + q, p - q, p * q, q / p)
print(p ** 2)
print(q // p, q % p)       # floor divide and remainder
print(np.add(p, q), np.subtract(p, q), np.multiply(p, q), np.divide(q, p))
```

```python
v = np.array([1.234, -2.567, 3.5, 4.49])
print(np.sqrt([1, 4, 9]))
print(np.square(p))
print(np.power(p, 3))
print(np.exp(p))          # e^x
print(np.log(p))          # natural log
print(np.log10([10, 100]))
print(np.log2([2, 8]))
print(np.abs(v))
print(np.round(v, 1))
print(np.floor(v), np.ceil(v), np.trunc(v))
print(np.sign(v))
print(np.clip(v, 0, 3))   # limit values between 0 and 3
```

```python
angles = np.array([0, np.pi/2, np.pi])
print(np.sin(angles))
print(np.cos(angles))
print(np.tan(angles))
print(np.degrees(np.pi))
print(np.radians(180))
```

## 6. Statistics

```python
s = np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(s), np.sum(s, axis=0), np.sum(s, axis=1))  # axis 0 = columns, axis 1 = rows
print(np.mean(s))
print(np.median(s))
print(np.std(s))        # standard deviation
print(np.var(s))        # variance
print(np.min(s), np.max(s))
print(np.argmin(s), np.argmax(s))   # position of min / max
print(np.ptp(s))        # max - min
print(np.percentile(s, 50))
print(np.quantile(s, 0.25))
print(np.cumsum(s))     # running total
print(np.cumprod(s))    # running product
print(np.prod(s))
```

```python
data1 = np.array([1, 2, 3, 4, 5])
data2 = np.array([2, 4, 6, 8, 10])
print(np.corrcoef(data1, data2))   # correlation
print(np.cov(data1, data2))        # covariance
print(np.average(data1, weights=[1, 1, 1, 1, 5]))
```

## 7. Sorting, Searching, Unique

```python
z = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print(np.sort(z))
print(np.argsort(z))         # indexes that would sort the array
print(np.unique(z))
print(np.unique(z, return_counts=True))
print(np.where(z > 3))              # positions where condition is True
print(np.where(z > 3, 'big', 'small'))
print(np.searchsorted(np.sort(z), 4))
print(np.flip(z))            # reverse
print(np.nonzero(z > 3))
```

## 8. Combining and Splitting

```python
a1 = np.array([[1, 2], [3, 4]])
a2 = np.array([[5, 6], [7, 8]])
print(np.concatenate([a1, a2], axis=0))  # stack down
print(np.concatenate([a1, a2], axis=1))  # stack side by side
print(np.vstack([a1, a2]))
print(np.hstack([a1, a2]))
print(np.stack([a1, a2]).shape)
print(np.split(np.arange(6), 3))
print(np.array_split(np.arange(7), 3))
print(np.tile([1, 2], 3))
print(np.repeat([1, 2], 3))
print(np.append(p, [7, 8]))
print(np.insert(p, 1, 99))
print(np.delete(p, 0))
```

## 9. Comparison and Logic

```python
c = np.array([1, 2, 3, 4])
print(c > 2)
print(np.any(c > 3), np.all(c > 0))
print(np.logical_and(c > 1, c < 4))
print(np.logical_or(c < 2, c > 3))
print(np.logical_not(c > 2))
print(np.isin(c, [2, 4]))
print(np.array_equal(c, [1, 2, 3, 4]))
print(np.allclose(0.1 + 0.2, 0.3))
print(np.count_nonzero(c > 1))
```

## 10. Random Numbers

```python
np.random.seed(42)                       # same results every time
print(np.random.rand(3))                 # 0 to 1
print(np.random.randn(3))                # normal distribution
print(np.random.randint(1, 100, 5))      # random integers
print(np.random.choice([10, 20, 30], 2)) # pick randomly
print(np.random.normal(50, 10, 3))       # mean 50, std 10
print(np.random.uniform(0, 5, 3))
arr = np.arange(5)
np.random.shuffle(arr)
print(arr)
print(np.random.permutation(5))

# Newer style (recommended)
rng = np.random.default_rng(42)
print(rng.integers(1, 10, 5))
print(rng.random(3))
```

## 11. Linear Algebra (Matrices)

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(np.dot(A, B))          # matrix multiplication
print(A @ B)                 # same thing
print(np.matmul(A, B))
print(np.linalg.det(A))      # determinant
print(np.linalg.inv(A))      # inverse
print(np.linalg.eig(A))      # eigenvalues and eigenvectors
print(np.linalg.norm(A))
print(np.linalg.matrix_rank(A))
print(np.trace(A))
print(np.diag(A))
print(np.linalg.solve(A, [5, 6]))   # solve Ax = b
print(np.outer([1, 2], [3, 4]))
print(np.inner([1, 2], [3, 4]))
print(np.cross([1, 0, 0], [0, 1, 0]))
```

## 12. Missing Values (NaN)

```python
n = np.array([1, 2, np.nan, 4])
print(np.isnan(n))
print(np.nansum(n), np.nanmean(n), np.nanmax(n), np.nanmin(n))
print(np.nan_to_num(n, nan=0))
print(n[~np.isnan(n)])      # remove NaN
print(np.isinf([1, np.inf]))
```

## 13. Broadcasting

```python
# NumPy can match different shapes automatically
g = np.array([[1, 2, 3], [4, 5, 6]])
print(g + 10)                 # add 10 to every item
print(g + np.array([1, 0, 1]))  # add row to every row
print(g * np.array([[2], [3]])) # multiply each row differently
```

## 14. Copy vs View

```python
orig = np.array([1, 2, 3])
view = orig[:]          # shares memory
copy = orig.copy()      # separate data
view[0] = 100
print(orig)   # changed!
print(copy)   # not changed
```

## 15. Saving and Loading

```python
np.save('my_array.npy', A)             # save one array
print(np.load('my_array.npy'))
np.savez('many.npz', a=A, b=B)         # save several
loaded = np.load('many.npz')
print(loaded['a'])
np.savetxt('data.csv', A, delimiter=',')
print(np.loadtxt('data.csv', delimiter=','))
```

## 16. Other Useful Functions

```python
print(np.diff([1, 4, 9, 16]))         # difference between neighbours
print(np.gradient([1, 4, 9, 16]))
print(np.histogram([1, 2, 2, 3, 3, 3], bins=3))
print(np.bincount([0, 1, 1, 2, 2, 2]))
print(np.digitize([1, 5, 8], bins=[0, 3, 6]))
print(np.meshgrid([1, 2], [3, 4]))
print(np.roll([1, 2, 3, 4], 1))
print(np.convolve([1, 2, 3], [0, 1]))
print(np.intersect1d([1, 2, 3], [2, 3, 4]))
print(np.union1d([1, 2], [2, 3]))
print(np.setdiff1d([1, 2, 3], [2]))
print(np.trapz([1, 2, 3]))
print(np.vectorize(lambda x: x * 2)(np.array([1, 2, 3])))
```
