####      Dictionary
Key-Value pair mapping container defined withcurly braces {  key : value }

Mutability : Mutable :  Values and Key-Value pair can be modified

Indexing & Slicing :    numeri indexing and sequence slicing are not allowed

Key-Based Indexing :    values are accessed via their key ( e.g .., dict [ "key" ] )

Heterogeneous Data :    Allowed Both key and values can hold multiple data types

Duplicates :            Keys cannot be duplicated (new value overwrite existing keys). Value can be duplicated

```python
student_data = {
                "Name" : ['jhon', 'jay', 'sai', 'bala'],
                "Age"  : [23, 24, 21, 45],
                "Department" : ['CSC', 'EEE', 'IT', 'ECE']
}
print(student_data)
print("1 :",student_data['Name'])      # get values from respective keys
print("2 :",student_data['Age'])
print("3 :",student_data['Department'])

# NOTE : dont put duplicate keys  duplicate values allowed last updates values shows
student_data['Id'] = ['id1','id2','id3','id4']  # Add new key values in dict
student_data['Name'][2] = 'ram' # update exist value using respective key name
#student_data['Name'] = 'ram' # update exist value using respective key name
print("4 :",student_data['Name']) 
print(student_data)

del student_data['Id']  # delete specified key



for key in student_data:    #   accessing dictionary key values using for loop
    print(key,':',student_data[key]) 
```

#### dict.items() -> (method) def items() -> dict_items[str, Any]
D.items() -> a set-like object providing a view on D's items

```python
employee_data_list = Employee_data.items()
print(employee_data_list)
print("Type of employee_data_list : ",type(employee_data_list))
print("Employee_data : ",type(Employee_data))
```

#### dict.copy -> () -> dict[str, Any] -> Return shallow copy of dictionay

D.copy() -> a shallow copy of D

```python
employee_data_copy = Employee_data.copy()  # return shallow copy of dict
print(employee_data_copy)
print("Type of employee_data_copy : ",type(employee_data_copy))
```

####    dict.keys( )  -> (method) def keys() -> dict_keys[str, Any]
D.keys() -> a set-like object providing a view on D's keys

```python
employee_data_list_keys = Employee_data.keys()  # return all dict keys
print(employee_data_list_keys)
print("Type of employee_data_list_keys : ",type(employee_data_list_keys))
```

#### dict.values( )  ->  (method) def values() -> dict_values[str, Any]
D.values() -> an object providing a view on D's values

```python
employee_data_list_values = Employee_data.values()  # return all dict values
print(employee_data_list_values)
print("Type of employee_data_list_values : ",type(employee_data_list_values))
```

#### dict.get() ->  Return the value for key if key is in the dictionary, else default.
(key: str, default: None = None, /) -> (Any | None)


```python
employee_data_get = Employee_data.get('Name')  # return values of specified key
print(employee_data_get)
print("Type of employee_data_get : ",type(employee_data_get))
```

####    dict.pop() -> remove specified key and return the corresponding value.
(method) def pop(
    key: str,
    /
) -> Any
D.pop(k[,d]) -> v, remove specified key and return the corresponding value.

If the key is not found, return the default if given; otherwise, raise a KeyError.

```python
employee_data_1 = Employee_data.copy() # take copy because original diction affect
print("1 :",employee_data_1)
employee_data_1.pop('Name')  # Note: in-place -> remove specified key and return the corresponding value.
print("2 :",employee_data_1)
print("Type of employee_data_1 : ",type(employee_data_1))

print("3 :",Employee_data)    # original dict not affect
```

####    dict.popitem()  -> (method) def popitem() -> tuple[str, Any]

```python
employee_data_1 = Employee_data.copy() # take copy because original diction affect


print("1 :",employee_data_1.popitem())  #  return : remove end (key:value) pairs

print("2 :",employee_data_1)

print("Type of employee_data_1 : ",type(employee_data_1))

print("3 :",Employee_data)    # original dict not affect

pd.DataFrame(employee_data_1)
```
