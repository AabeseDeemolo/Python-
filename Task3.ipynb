Rock, Paper, Scissors Game
Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
 This task will test your understanding of:
●	
Input/output 
●	
Conditional logic
●	
Random module
●	
Loops
●	
Exception handling

Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.

```python
import random as rd



def you_want_play_again_msg():      # play again option selection function
    while True:
        try:
            value = input("Enter if 'Yes' Play Again \n Else Enter 'No to Quite " ).strip().lower()  ## avoid unwanted space and case problems
            if value == "yes":
                game_start()
                break
            elif value!= "yes" and value != "no":
                continue
            elif value == "no": 
                print(" You Exit !!!") 
                return False
        except ValueError as e:
            print("Error : ",e)
            continue
                           
###############################################################################################################################

def rock_paper_siser_game(user_choice):             ## Game logig function
    
    computer_choice = str(rd.choice(["Rock","Paper","Siser"])).lower()      # randomly select optioins
    
    def computer_wins_message():                            # computer wins message function
        print(f"I Choose {computer_choice} You Choose {user_choice}")
        print("!!! COMPUTER WON !!!")
        you_want_play_again_msg()         # play again function call

    def player_wins_message():                              # player wins message function
        print(f"I Choose {computer_choice} You Choose {user_choice}")
        print("!!! YOU WON !!!")
        you_want_play_again_msg()         # play again function call
            

    if computer_choice == "rock" and user_choice == "siser":
        computer_wins_message()
    elif computer_choice == "siser" and user_choice == "rock":
        player_wins_message()
    elif computer_choice == "rock" and user_choice == "paper":
        player_wins_message()
    elif computer_choice == "paper" and user_choice == "rock":
        computer_wins_message()
    elif computer_choice == "paper" and user_choice == "siser":
        player_wins_message()
    elif computer_choice == "siser" and user_choice == "paper":
        computer_wins_message()

    else:
        print(f"I Choose {computer_choice} You Choose {user_choice}")
        print("!!! Tie !!!")
        you_want_play_again_msg()
        
#######################################################################################################################     





def game_start():       # game start function
    
    print("Now You Will Play 'Rock' 'Paper' 'Siser \n If You Want Quite Enter 'Q' ")
    while True:
        try:
            user_choice = input("Enter Your Choice : ").strip().lower()  # avoid unwanted space and casesensitive problems

            if user_choice == "q":          # logic if player wants Quit 
                print(" You Exit !!!")
                break

            elif user_choice not in str(["rock","paper","siser"]).lower():      # check player enter correct option
                print("Invalid Input : You Input Is Wrong Enter 'ROCK' or 'PAPER' or 'SISER' ")
                continue

            else: 
                rock_paper_siser_game(user_choice)
                break

        except ValueError as e:
            print("Error : ",e)
            print("""Now You Will Play 'Rock' 'Paper' 'Siser
            If You Want Quite Enter  Q """)
            

game_start()  # Game Start function call
```
