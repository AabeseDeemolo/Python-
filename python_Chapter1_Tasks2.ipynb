#### 2.5 . user defined functions without for loop, if loop

```python
# Addition of two numbers
try:
    num_1 = int(input("Enter num 1 : "))
    num_2 = int(input("Enter num 2 : "))

    def add(num_1,num_2):
        return num_1+num_2
    
    print(add(num_1,num_2))
    
except ValueError as e:
    print(e)

```

```python

# Division of two numbers
try:
    num_1 = int(input("Enter num 1 : "))
    num_2 = int(input("Enter num 2 : "))

    def div(num_1,num_2):
        return num_1/num_2
    
    print(div(num_1,num_2))
    
except ValueError as e:
    print(e)
except ZeroDivisionError as e:
    print(e)

```

#### 2.6 . user defined functions with for loop

```python
# function for print numbers 1 to n
try:
    num = int(input("Enter num 1 to n : "))

    def print_number(num):
        for i in range(1,num+1):
            print(i)
    print_number(num)
    
except ValueError as e:
    print(e)


```

```python
# sum of n  numbers 1 to n
try:
    num = int(input("Enter num 1 to n : "))

    def sum_of_n_numbers(num):
        result = 0 
        for i in range(1,num+1):
            result = result + i  
        return result
    
    print(sum_of_n_numbers(num))
    
except ValueError as e:
    print(e)

```

#### 2.7 . user defined functions with for loop and if condition

```python
# Factors of  no program
try:
    num = int(input("Enter number  : "))
    
    def factors_of_num(num):
        for i in range(2,num):
            if num % i == 0:
                print(i)
    
    factors_of_num(num)
    
except ValueError as e:
    print(e)

```

```python
# check continuesly number is even
try:
    num = int(input("Enter number  : "))
    
    def check_even(num):
        for i in range(2,num+1):
            if i % 2 == 0:
                print(i,"is Even")
    
    check_even(num)
    
except ValueError as e:
    print(e)

```

#### 2.8  .  Break

```python
# use Break when fullfill the condition imediately terminate the loop

try:
    num = int(input("Enter breaking point number 1 to 100 : "))
    
    for i in range(1,101):
        if num == i:
            break
        else:
            print(i)

    
except ValueError as e:
    print(e)

```

```python
# breaks when find a square of the given number
try:
    num = int(input("Enter a number : "))
    
    for i in range(1,101):
        print(i)
        if i * i == num :
            break
        

    
except ValueError as e:
    print(e)

```

#### 2.9 . continue 
skip the current iteration of the given num goto next iteration

```python
# skip the current iteration of the given num goto next iteration
try:
    num = int(input("Enert a number 1 to 20 you dont want : "))
    
    for i in range(1,20):
        
        if i == num:
            continue
        else:
            print(i)
        
except ValueError as e:
    print(e)
```

```python
# skip the even numbers
try:
    
    for i in range(1,20):
        
        if i % 2 == 0:
            continue
        else:
            print(i)
        

    
except ValueError as e:
    print(e)
```
