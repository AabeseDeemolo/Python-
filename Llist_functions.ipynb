#### list Index Access

```python
empty_list = []             # empty list Creation 

heterogeneous_list = ['A','a','Abc',1,123,4.567,'A',"!@$*"]         # list with names

print("1 : List values : ",heterogeneous_list)
print("2 : Index[0] : ",heterogeneous_list[0])          # INDEXING : that returnes the respective value from index

print("3 : change value using index[0]")
heterogeneous_list[0] = "Abraham"     # MUTABLE : change the value using index
print("4 :", heterogeneous_list)

print("5 :  Get values using index")
print("6 :",heterogeneous_list[7])                                       
print("7 :", heterogeneous_list[0])                                        
print("8 :", heterogeneous_list[6])     #   list allowes duplicates                                   



try:
   heterogeneous_list['Rakchitha'] = "Ratha"     # TypeError: list indices must be integers or slices, not str
   print(heterogeneous_list[11])                 # IndexError : list index out of range

except IndexError as e:
   print(e)
except TypeError as e:
   print(e)

```

#### List Slicing [ start : end-1 : step ]

```python
#  list[ start : end - 1 : step]  -> if end is 8 thats make 6

num_list = [0,1,2,3,4,5,6,7]          # use numbers of tuple for easy understand slice concepts


print("length : ",len(num_list))
print("1 :",num_list[0:])     # starts from 0 th position
print("2 :",num_list[0:8])    # starts from  0 th position ends with 8-1 th position
print("3 :",num_list[0:8:2])  # step include 
print("4 :",num_list[0:8])
print("5 :",num_list[2:8])
print("6 :",num_list[1:8:3])

print("\n   Reverse slincing")          #---------   Nagative Indexing Allowed   -------------
print("7 :",num_list[:-1])     # start from 0 ends with backward -1  
print("8 :",num_list[-8:-4])     # starting index and ending index nagative (backward)
print("9 :",num_list[-6:-1:2])     # nagative indexes with step

print("\n   Reverse a string use slicing")
print("10 :",num_list[::-1])     # reverse the list using slice
print("11 :",num_list[::-2])      # reverse the list step by 



```

#### 2.1 .  list.append() -> Append object to the end of the list.
(method) def append(
    object: Any,
    /
) -> None

```python

fruits = ["apple","orange","banana","guva","pomogranate"]

fruits.append('grapes')     # append function add the value at end
print("1 : ",fruits)         
fruits.append('papaya')     # append function add the value at end
print("2 : ",fruits)
fruits.append(['mango','lemon'])    # appent list inside list nested list
print("3 :",fruits)
```

#### 2.2 .  list.pop() -> Remove and return item at index (default last).

Raises IndexError if list is empty or index is out of range.

```python

fruits = ['apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya']

try:
    print("1 : Index of removed value -> ",fruits.pop())     # function remove the last index value and return removed item
    print("2 : ",fruits)         
    print("3 : Index of removed value -> ",fruits.pop())     
    print("4 : ",fruits)
except IndexError as e:     # Raises IndexError if list is empty or index is out of range.
    print(e)
```

#### 2.3 .  list.copy() -> Return a shallow copy of the list.

(method) def copy() -> list[str]

```python
#   copy()

fruits = ['apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya']


fruits_copy = fruits.copy()
print("1 : fruits -> ",fruits)     #  Return a shallow copy of the list : use other variable to store
print("2 : fruits_copy -> ",fruits_copy)
  
```

#### 2.4 .  list.count() -> Return number of occurrences of value.
(method) def count( value: str, /) -> int


```python

fruits = ['apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya','guva','apple','guva']


print("1 : ",fruits)         
print("2 : Count of guva -> ",fruits.count("guva"))     # count of the specific value
print("3 : Count of apple -> ",fruits.count("apple"))     


```

#### 2.5 .  list.index() -> Return first index of value : Raises ValueError if the value is not present.
(method) def index(
    value: str,
    start: SupportsIndex = 0,
    stop: SupportsIndex = sys.maxsize,
    /
) -> int


```python

fruits = ['apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya','guva','apple','guva',['mango', 'lemon']]

try:
    print("1 : ",fruits)         
    print("2 : Index of apple -> ",fruits.index("apple"))     
    print("3 : Index of pomogranate -> ",fruits.index("pomogranate"))     # return the first occurence of the value
    print("4 : Index of values in sub list -> ",fruits.index("guva",0,9))     # return the first occurence of the specified value with in start to end (sub list) 
    print("5 : Index of values in sub list -> ",fruits.index("guva",4,9))     # return the first occurence of the specified value with in start to end (sub list) 
    print("6 :",fruits[10])  # nested list indexing
    print("7 :",fruits[10][1])
except ValueError as e:     # Raises ValueError if the value is not present
    print(e)
```

#### 2.6 .  list.extend() -> Extend list by appending elements from the iterable.
(method) def extend(
    iterable: Iterable[str],
    /
) -> None

```python

fruits = ['apple', 'orange', 'banana']


print("1 : ",fruits)         
fruits.extend("mango")   # iterates each character 
print("2 : ",fruits) 
fruits.extend(['guvi','watermelon'])    # iterates each character and append end of list as a single list , NOTE: tuples and dict and str data types are
print("3 : ",fruits)
fruits.extend(('mango','lemon'))        # tuples of values -> ('mango','lemon')
print("4 : ",fruits)
fruits.extend({'grapes' : 3,'lemon' : 5})        # dict keys only accept -> {'grapes' : 3,'lemon' : 5}
print("5 : ",fruits)
    
```

#### 2.8 . list.remove() - > Remove first occurrence of value  ->   Raises ValueError if the value is not present.

#### 2.9 . list.sort() -> Sort the list in ascending order and return None.

The sort is in-place (i.e. the list itself is modified) and stable (i.e. the order of two equal elements is maintained).

If a key function is given, apply it once to each list item and sort them, ascending or descending, according to their function values.

The reverse flag can be set to sort in descending order.

(method) def sort(
    *,
    key: None = None,
    reverse: bool = False
) -> None


```python

fruits = ['apple','orange','banana','guva','pomogranate','grapes','papaya','guva','apple','guva']
numbers = [5,6,7,1,2,3,4,8,9,0]


print("1 : ",fruits)
fruits.sort()  # sort values by Asending order
print("2 :  sorted Acend -> ",fruits)
fruits.sort(reverse=True)      # sort values by Desending (reverse order)
print("3 :  sorted Dcend -> ",fruits)
numbers.sort()
print("4 :  sorted Acend -> ",numbers)
numbers.sort(reverse=True)  # sort values by Desending (reverse order)
print("5 :  sorted Dcend -> ",numbers)
```

#### 2.10 . list.reverse() -> reverse list Inplace

```python

fruits = ['apple','orange','banana','guva','pomogranate','grapes','papaya','guva','apple','guva']
numbers = [5,6,7,1,2,3,4,8,9,0]


print("1 : ",fruits)
fruits.reverse()  # reverse the list
print("2 :  reversed -> ",fruits)


print("3 :  ",numbers)
numbers.reverse()  # sort values by Desending (reverse order)
print("4 :  reversed -> ",numbers)
```

####        tuple ( )

##### Sequence container defined with square brackets [ ]

Immutable -> onces created cannot be altered,added, or removed

Indexing & Slicing : Allowed -> position based Index, support standard slicing operation

Heterogeneous Data : Allowed -> Store any mix of data

Duplicates : Allowed -> identical elements can be stored without restriction

```python
empty_tuple = ( )             # empty tuple Creation 

heterogeneous_tuples = ('A','a','Abc',(1,2,3),123,4.567,'A',"!@$*")         # tuples with names

print("1 : tuple values : ",heterogeneous_tuples)
print("2 : Index[0] : ",heterogeneous_tuples[0])          # INDEXING : that returnes the respective value from index

print("3 : change value using index[0]")
print("4 :", heterogeneous_tuples)

print("5 :  Get values using index")
print("6 :",heterogeneous_tuples[7])                                       
print("7 :", heterogeneous_tuples[0])                                        
print("8 :", heterogeneous_tuples[6])     #   tuples allowes duplicates   
print("\n nested tuples")
print("9 : Index[3][0] :", heterogeneous_tuples[3][0])     #   nested tuples : tule inside tuple  
print("10 : Index[3][1]", heterogeneous_tuples[3][1])     #          



print("\nError Handeling")
try:
   heterogeneous_tuples['Rakchitha'] = "Ratha"     # TypeError: 'tuple' object does not support item assignment
   print(heterogeneous_tuples[11])                 # IndexError : tuples index out of range

except IndexError as e:
   print(e)
except TypeError as e:
   print(e)

```

```python
#  tuple[ start : end - 1 : step]  -> if end is 8 thats make 6

num_tuple = (0,1,2,3,4,5,6,(1,2,3))          # i use numbers of tuple because easy understand
print("Type : ",type(num_tuple))

print("length : ",len(num_tuple))
print("1 :",num_tuple[0:])     # starts from 0 th position
print("2 :",num_tuple[0:8])    # starts from  0 th position ends with 8-1(end-1) th position
print("3 :",num_tuple[0:8:2])  # step include 
print("4 :",num_tuple[0:8])
print("5 :",num_tuple[2:8])
print("6 :",num_tuple[1:8:3])

print("\n   Reverse slincing")          #---------   Nagative Indexing Allowed   -------------
print("7 :",num_tuple[:-1])     # start from 0 ends with backward -1  
print("8 :",num_tuple[-8:-4])     # starting index and ending index nagative (backward)
print("9 :",num_tuple[-6:-1:2])     # nagative indexes with step

print("\n   Reverse a string use slicing")  #-----------    Reverse tuple   ---------------#
print("10 :",num_tuple[::-1])     # reverse the tuple using slice
print("11 :",num_tuple[::-2])      # reverse the tuple step by 



```
