####        set { }

#####   Uordered collection of unique items defined with curiy braces { }  or set( )
Mutability : Mutable -> add or remove elements

Indexing & Slicing : Not Allowed -> Items have no fixed positions

Heterogeneous Data : Allowed -> Store any mix of data 

Duplicates : Not Allowed -> Automatically removes duplicates

```python
empty_set = set()            # empty set Creation 
print("empty set : ",empty_set)

heterogeneous_set = {'A','a','Abc',1,123,4.567,'A',"!@$*"}         # list with names

print(" set values : ",heterogeneous_set)


```

(element: Any, /) -> None

#### 3 . set.pop( )     ->  Remove first element in the set

#### 4 . set.remove( )  ->  Remove an element from a set; it must be a member.If the element is not a member, raise a KeyError.


(method) def remove(
    element: str,
    /
) -> None


```python
fruits = {'apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya','guva','apple','guva'}     # set with duplicate values

print("1 : fruits :",fruits)
fruits.remove('apple')     # remove specific element if present
print("2 : After remove apple from fruits :",fruits)

try:
    fruits.remove('lemon')     # remove unpresent member in list
except KeyError as e:
    print(f"3 : Error Message : {e} not present in set")
    pass
```

#### 5 . set.discard( )  ->   Remove an element from a set if it is a member. If the element is not a member, do nothing.
(method) def discard(
    element: object,
    /
) -> None


```python
fruits = {'apple', 'orange', 'banana', 'guva', 'pomogranate', 'grapes', 'papaya','guva','apple','guva'}     # set with duplicate values

print("1 : fruits :",fruits)
fruits.discard('apple')     # remove specific element if present
print("2 : After remove apple from fruits :",fruits)
fruits.discard('lemon')     # this member not present in the set. so do nothing NOTE: but remove function throw Error
print("3 : fruits :",fruits)
```

#### 5 . set.update( )  ->   Update a set with the union of itself and others.
(method) def update(*s: Iterable[str]) -> None


```python
fruits = {'apple', 'banana', 'guva', 'pomogranate', 'grapes','guva','apple','guva'}     # set with duplicate values

print("1 : fruits :",fruits)
fruits.update(['lemon','watermelon'])     # update list of elements into set
print("2 : fruits :",fruits,"-> update list of elements into set")
fruits.update(('papaya','orange'))     # update tuples of elements into set
print("3 : fruits :",fruits,"-> update tuples of elements into set")
fruits.update({'guvi','mango'})     # update set of elements into set
print("4 : fruits :",fruits,"-> update set of elements into set")

```

####    7 . set.difference_update( )    ->  Remove all elements of another set from this set.
(method) def difference_update(*s: Iterable[object]) -> None


```python

fruits_1 = {'apple', 'banana', 'pomogranate', 'grapes','guvi','apple','guva'}     # set with duplicate values
fruits_2 = {'apple','papaya','guvi','orange','mango','guva'}

print("1 : fruits_1 :",fruits_1)
print("2 : fruits_2 :",fruits_2)

fruits_1.difference_update(fruits_2)        # NOTE : UPDATE in-place
print("3 : fruits_1.difference_update(fruits_2) -> :",fruits_1)    # Remove all elements of another set from this set.
fruits_2.difference_update(fruits_1)
print("4 : fruits_2.difference_update(fruits_1) -> :",fruits_2)    # Remove all elements of another set from this set.



```

#### 10 . set.isdisjoint( ) -> Return True if two sets have a null intersection
(method) def isdisjoint(
    s: Iterable[object],
    /
) -> bool


```python

fruits_1 = {'apple', 'banana', 'pomogranate', 'grapes','guvi','apple'}    
fruits_2 = {'papaya','orange','mango','guva'}
fruits_3 = {'apple','papaya','guvi','orange','mango','guva'}

print("1 : fruits_1 :",fruits_1)
print("2 : fruits_2 :",fruits_2)
print("2 : fruits_3 :",fruits_3)

print("3 : fruits_1.isdisjoint(fruits_2) -> ",fruits_1.isdisjoint(fruits_2))    # Return True : any one element not present in both sets
print("4 : fruits_2.isdisjoint(fruits_3) -> ",fruits_2.isdisjoint(fruits_3))    # Return False :  any one element present in both sets



```
